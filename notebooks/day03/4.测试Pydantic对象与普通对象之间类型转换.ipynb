```python
from dataclasses import dataclass
from pydantic import BaseModel, ConfigDict

# Pydantic 响应模型：通过 from_attributes=True 允许从普通对象实例中提取字段值
class UserResponse(BaseModel):
    model_config = ConfigDict(from_attributes=True)
    id: int
    name: str
    age: int

# 普通 Python 类，和 Pydantic 模型并无直接继承关系
# class User:
#     def __init__(self, id, name, age):
#         self.id = id
#         self.name = name
#         self.age = age

@dataclass
class User:
    id:int
    name: str
    age: int
    phone:str

# 创建一个普通对象实例，用于演示对象转 Pydantic 模型
user = User(1, "jack", 19,"153234647")

# model_validate 会按属性名自动映射普通对象中的字段到模型中
user_response = UserResponse.model_validate(user)
print(type(user_response))
print(user_response)
```
