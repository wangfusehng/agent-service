```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage, AIMessage
from langgraph.checkpoint.memory import InMemorySaver

load_dotenv()

agent = create_agent(
    model="deepseek-v4-flash",
    checkpointer=InMemorySaver()
)

#设置会话标识,要能查询记忆，必须携带thread_id
config = {"configurable":{"thread_id":"thread_1"}}
messages = {
    "messages":[
        HumanMessage("你好，我是程序员福生"),
        AIMessage("hello,我是deepseek,有什么可以帮助你。"),
        HumanMessage("介绍一下希尔排序")
    ]
}
messages1 = {
    "messages":[
         HumanMessage("我是谁？")
    ]
}
response = agent.invoke(messages,config)
response1 = agent.invoke(messages1,config)
print(response["messages"][-1].content)
```

```python
print(response1["messages"][-1].content)
```

```python
snapshot = agent.get_state(config=config)
for message in snapshot.values["messages"]:
    # message.pretty_print()
    if isinstance(message, HumanMessage):
        role = "human"
    elif isinstance(message, AIMessage):
        role = "assistant"
    else:
        continue
    print({"role":role,"content":message.content})
```
