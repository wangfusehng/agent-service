```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field
load_dotenv()
#1.读取环境变量
#2.初始化模型
model = init_chat_model(model="deepseek-v4-flash",extra_body={"thinking":{"type":"disabled"}})
#3.定义Pydantic BaseModel类，用于约束模型输出结果
class StudentInfo(BaseModel):
    name: str=Field(description="The name of this student")
    age:int=Field(description="The age of this student")
    gender:str=Field(description="The gender of this student")
    address:str=Field(description="The address of this student")
    hobby:str=Field(description="The hobby of this student")
#4.用模型绑定BaseModel
model_with_structured_output = model.with_structured_output(schema=StudentInfo)
#5.调用模型，完成结构化输出
response = model_with_structured_output.invoke("小明今年15岁，喜欢打篮球，家住浙江杭州")
response
```
