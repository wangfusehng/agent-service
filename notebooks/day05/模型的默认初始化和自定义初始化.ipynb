```python
from dashscope.applications.finetune.reinforcement import DASHSCOPE_API_KEY
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.tools import tool
#1.加载环境变量
load_dotenv()
#2.定义工具
@tool
def get_weather(location:str)->str:
    """
     Get weather of a location, the user should supply a location first.
    :param location:The city and state, e.g. San Francisco, CA
    :return: The weather of given location
    """
    return f"Current weather in {location} is sunny"
#3.定义模型
model = init_chat_model(model="deepseek-v4-flash",extra_body={"thinking":{"type":"disabled"}})
#4.初始化agent
myagent = create_agent(model=model,tools=[get_weather])
#5.调用agent
messages = {"messages":[{"role":"user","content":"北京今天天气如何？"}]}
response = myagent.invoke(messages)
print(response)
```

```python
from dotenv import load_dotenv
import os
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.tools import tool
#1.加载环境变量
load_dotenv()

#2.定义工具
@tool
def get_weather(location:str)->str:
    """
     Get weather of a location, the user should supply a location first.
    :param location:The city and state, e.g. San Francisco, CA
    :return: The weather of given location
    """
    return f"Current weather in {location} is sunny"
#3.定义模型,自定义初始化，把Langchain不支持的模型伪装成OpenAI,但是要自己传aip_key和base_url
model = init_chat_model(model="qwen4-max",
                        model_provider="openai",
                        api_key=os.getenv("DASHSCOPE_API_KEY"),
                        base_url=os.getenv("DASHSCOPE_BASE_URL"),
                        )
#4.初始化agent
myagent = create_agent(model=model,tools=[get_weather])
#5.调用agent
messages = {"messages":[{"role":"user","content":"北京今天天气如何？"}]}
response = myagent.invoke(messages)
print(response)
```

```python
# 某些模型提供商并不兼容OpenAI规范，也不被init_chat_model默认初始化支持，就必须用专用的Model类来初始化了，如Ollama

from langchain_ollama import ChatOllama
#默认访问本地的模型，无需传递api_key和base_url
model = ChatOllama(
    model="qwen3.5:4b"
)
@tool
def get_weather(location:str)->str:
    """
     Get weather of a location, the user should supply a location first.
    :param location:The city and state, e.g. San Francisco, CA
    :return: The weather of given location
    """
    return f"Current weather in {location} is sunny"
agent = create_agent(model=model,tools=[get_weather])
messages = {"messages":[{"role":"user","content":"北京今天天气如何？"}]}
response = myagent.invoke(messages)
print(response)
```
